```python
import requests
```

```python
# 1. On prépare l'adresse
url_gps = "https://nominatim.openstreetmap.org/search?q=Paris,France&format=json"

# 2. LA NOUVEAUTÉ : On met un masque à notre robot pour ne pas être bloqué
mes_entetes = {"User-Agent": "MonProjetKayak/1.0"}

# 3. On envoie la requête AVEC notre masque
reponse = requests.get(url_gps, headers=mes_entetes)

# 4. On ouvre le JSON (cette fois, le serveur nous l'accepte)
donnees = reponse.json()[0]

print("La Latitude de Paris est :", donnees["lat"])
print("La Longitude de Paris est :", donnees["lon"])
print("L'ID unique de la ville est :", donnees["place_id"])
```

```python
import os
import requests

# Test 1 : Vérifier la librairie dotenv
try:
    from dotenv import load_dotenv
    print("✅ ETAPE 1 : L'outil 'dotenv' est bien installé et reconnu !")
except ImportError:
    print("❌ ERREUR ETAPE 1 : L'outil 'dotenv' est introuvable.")

# Test 2 : Vérifier la clé secrète
load_dotenv()
api_key = os.environ.get("OPENWEATHER_API_KEY")

if not api_key:
    print("❌ ERREUR ETAPE 2 : La clé API n'a pas été trouvée dans le fichier .env")
else:
    print("✅ ETAPE 2 : La clé secrète a bien été extraite du coffre-fort !")
    
    # Test 3 : Vérifier l'API Météo
    lat = 48.8534
    lon = 2.3483
    url = f"https://api.openweathermap.org/data/3.0/onecall?lat={lat}&lon={lon}&exclude=current,minutely,hourly,alerts&units=metric&appid={api_key}"
    
    print("⏳ ETAPE 3 : Envoi de la requête au serveur météo...")
    reponse = requests.get(url)
    donnees = reponse.json()
    
    if "daily" in donnees:
        print("✅ ETAPE 3 REUSSIE : La météo fonctionne ! Demain il fera", donnees["daily"][0]["temp"]["day"], "°C")
    else:
        print("❌ ERREUR ETAPE 3 : Le serveur a refusé de donner la météo. Voici son message d'erreur officiel :")
        print(donnees)
```
